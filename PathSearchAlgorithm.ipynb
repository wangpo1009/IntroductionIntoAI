#  Some Path Searching Algorithm
In this session, we will study about how we store graphs and BFS, DFS, UCS algorithms 

## Reading the input 
- Our graph input is going to be a distance matrix which will be store as an adjacency dict



```python
from collections import defaultdict
from queue import Queue, PriorityQueue

# Read the input from txt file
def read_txt(file):
    size = int(file.readline())
    start, goal = [int(num) for num in file.readline().split(' ')]
    matrix = [[int(num) for num in line.split(' ')] for line in file]
    return size, start, goal, matrix

# Convert adjacency matrix into adjency dict
def convert_graph(a):
    adjList = defaultdict(list)
    for i in range(len(a)):
        for j in range(len(a)):
            if a[i][j] == 1:
                adjList[i].append(j)
    return adjList

def convert_graph_weight(a):
    adjList = defaultdict(list)
    for i in range(len(a)):
        for j in range(len(a)):
            if a[i][j] != 0:
                adjList[i].append((j, a[i][j]))
    return adjList

```

## BFS Implementation 

```python
# Original code
def BFS(graph, start, end):
    visited = []
    frontier = Queue()
    # thêm node start vào frontier và visited
    frontier.put(start)
    visited.append(start)
    # start không có node cha
    parent = dict()
    parent[start] = None
    path_found = False
    while True:
        if frontier.empty():
            raise Exception("No way Exception")
        current_node = frontier.get()
        visited.append(current_node) #<---- ERROR 01
        # Kiểm tra current_node có là end hay không
        if current_node == end:
            path_found = True
            break
        for node in graph[current_node]:
            if node not in visited:
                frontier.put(node)
                parent[node] = current_node
                visited.append(node) #<---- ERROR 02
    # Xây dựng đường đi
    path = []
    if path_found:
        path.append(end)
        while parent[end] is not None:
            path.append(parent[end])
            end = parent[end]
        path.reverse()
    return path

# Testing
result_bfs = BFS(graph_1, start_1, goal_1)
print("Kết quả sử dụng thuật toán BFS: \n", result_bfs)
# -> No syntax error
```

### ERROR analysis
#### Syntax
- Không chứa lỗi syntax
#### ERROR 01: Thừa lệnh 
- Do ở bước sinh nút phía trên đã cho visited.append(node) rồi, nên việc gọi thêm visited.append(current_node) ở đây thể hiện sự không đồng nhất, đồng thời nếu ta sửa lại chuẩn (chỉ đánh dấu visited khi lấy nút ra khỏi frontier), thì vị trí này mới chính là vị trí đúng duy nhất cần đánh dấu.

#### ERROR 02: Đánh dấu visited quá sớm
- Ngay khi một nút con vừa được sinh ra (chưa hề được xét duyệt xem có phải là đích hay không), code đã lập tức gọi visited.append(node). Điều này khiến nút đó bị khóa lại, không bao giờ được duyệt lại từ các nhánh khác tối ưu hơn. Đối với thuật toán tìm kiếm đường đi ngắn nhất (như BFS), điều này làm mất đi tính tối ưu hoặc dẫn đến tìm sai đường đi ngắn nhất.


```python
def fixed_BFS(graph, start, end):
    visited = []
    frontier = Queue()
    
    # Chỉ đưa start vào frontier ban đầu
    frontier.put(start)
    
    parent = dict()
    parent[start] = None
    path_found = False
    
    while True:
        if frontier.empty():
            raise Exception("No path found") # Sửa lại tên exception cho hợp lý
            
        current_node = frontier.get()
        
        # KIỂM TRA ĐÍCH NGAY KHI LẤY RA KHỎI FRONTIER
        if current_node == end:
            path_found = True
            break
            
        # Đánh dấu visited KHI VÀ CHỈ KHI lấy nút ra xét (expand)
        if current_node not in visited:
            visited.append(current_node)
            
            for node in graph[current_node]:
                if node not in visited and node not in list(frontier.queue): # Hoặc kiểm tra cơ bản
                    frontier.put(node)
                    if node not in parent: # Đảm bảo giữ cha ngắn nhất cho BFS
                        parent[node] = current_node
                        
    # Xây dựng đường đi
    path = []
    if path_found:
        path.append(end)
        while parent[end] is not None:
            path.append(parent[end])
            end = parent[end]
        path.reverse()
    return path

# Testing
result_bfs = fixed_BFS(graph_1, start_1, goal_1)
print("Kết quả sử dụng thuật toán BFS sau sửa lỗi: \n", result_bfs)
```

## DFS Implementation

```python
# Original Code
def DFS(graph, start, end):
    visited = []
    frontier = []
    # thêm node start vào frontier và visited
    frontier.append(start)
    visited.append(start)
    # start không có node cha
    parent = dict()
    parent[start] = None
    path_found = False
    while True:
        if frontier == []:
            raise Exception("No way Exception")
        current_node = frontier.pop()
        visited.append(current_node) # <---- ERROR 01
        # Kiểm tra current_node có là end hay không
        if current_node == end:
            path_found = True
            break
        for node in graph[current_node]:
            if node not in visited:
                frontier.append(node)
                parent[node] = current_node
                visited.append(node) # <---- ERROR 02
    # Xây dựng đường đi
    path = []
    if path_found:
        path.append(end)
        while parent[end] is not None:
            path.append(parent[end])
            end = parent[end]
        path.reverse()
    return path

# Test
result_dfs = DFS(graph_1, start_1, goal_1)
print("Kết quả sử dụng thuật toán DFS: \n", result_dfs)
```

#### ERROR 01: Thừa lệnh 
- Tương tự với BFS

#### ERROR 02: Lỗi đánh dấu visited
- Tương tự với BFS

```python
def fixed_DFS(graph, start, end):
    visited = set()
    frontier = [] # Sử dụng list làm Stack (LIFO)
    
    frontier.append(start)
    parent = {start: None}
    path_found = False
    
    while len(frontier) > 0:
        current_node = frontier.pop()
        
        # Kiểm tra đích khi lấy ra khỏi frontier
        if current_node == end:
            path_found = True
            break
            
        # Chỉ đánh dấu visited khi chính thức xét (expand) nút đó
        if current_node not in visited:
            visited.add(current_node)
            
            # Duyệt các đỉnh kề và đưa vào frontier
            for node in graph[current_node]:
                if node not in visited:
                    frontier.append(node)
                    # Ghi nhận node cha (lưu ý: với DFS, nếu một node đã có cha rồi, 
                    # ta có thể ghi đè hoặc giữ nguyên tùy theo yêu cầu đường đi)
                    if node not in parent:
                        parent[node] = current_node
                        
    # Xây dựng đường đi từ đích về gốc
    path = []
    if path_found:
        curr = end
        while curr is not None:
            path.append(curr)
            curr = parent[curr]
        path.reverse()
    return path

# Test
result_dfs = fixed_DFS(graph_1, start_1, goal_1)
print("Kết quả sử dụng thuật toán DFS: \n", result_dfs)
```

## UCS Implementation

```python
# Original code
def UCS(graph, start, end):
    visited = []
    frontier = PriorityQueue()
    # thêm node start vào frontier và visited
    frontier.put((0, start))
    visited.append(start)
    # start không có node cha
    parent = dict()
    parent[start] = None
    path_found = False
    while True:
        if frontier.empty():
            raise Exception("No way Exception")
        current_w, current_node = frontier.get()
        visited.append(current_node) # <---- ERROR 01
        # Kiểm tra current_node có là end hay không
        if current_node == end:
            path_found = True
            break
        for nodei in graph[current_node]:
            node, weight = nodei
            if node not in visited:
                frontier.put((current_w + weight, node))
                parent[node] = current_node
                visited.append(node) # <---- ERROR 02
    # Xây dựng đường đi
    path = []
    if path_found:
        path.append(end)
        while parent[end] is not None:
            path.append(parent[end])
            end = parent[end]
        path.reverse()
    return current_w, path

# Test
result_ucs = UCS(graph_2, start_2, goal_2)
print("Kết quả sử dụng thuật toán UCS: \n", result_ucs)
```

### ERROR analysis
#### Syntax
- Không có lỗi syntax

#### ERROR 01: Lặp
- Tương tự DFS và BFS

#### ERROR 02: Đánh dấu ngay khi sinh nút
- Khi một đỉnh được phát hiện lần đầu bằng một đường đi dài, nó sẽ bị đưa vào visited và bị khóa lại. Sau đó, nếu có một đường đi ngắn hơn dẫn tới đỉnh đó, thuật toán sẽ bỏ qua vì tưởng rằng đỉnh đã được thăm. Điều này làm cho UCS mất đi tính tối ưu, tìm ra đường đi sai hoặc thậm chí không tìm được đường đi (báo lỗi No way Exception).

```python
def fixed_UCS(graph, start, end):
    visited = set()
    frontier = PriorityQueue()
    
    # Đưa tuple (chi phí, node) vào hàng đợi ưu tiên
    frontier.put((0, start))
    
    parent = {start: None}
    cost_so_far = {start: 0}  # Lưu chi phí nhỏ nhất tới mỗi node
    path_found = False
    final_cost = 0
    
    while not frontier.empty():
        current_w, current_node = frontier.get()
        
        # Nếu node đã được duyệt qua đường đi tối ưu trước đó thì bỏ qua
        if current_node in visited:
            continue
            
        # Đánh dấu visited KHI VÀ CHỈ KHI lấy ra khỏi frontier (expand node)
        visited.add(current_node)
        
        # Kiểm tra đích khi lấy ra khỏi frontier với chi phí tối ưu nhất
        if current_node == end:
            path_found = True
            final_cost = current_w
            break
            
        # Duyệt các đỉnh kề (node, weight)
        for node, weight in graph[current_node]:
            new_cost = current_w + weight
            
            # Nếu tìm thấy đường đi mới tốt hơn tới node (hoặc node chưa được thăm/chưa có trong cost_so_far)
            if node not in cost_so_far or new_cost < cost_so_far[node]:
                cost_so_far[node] = new_cost
                parent[node] = current_node
                frontier.put((new_cost, node))
                
    # Xây dựng đường đi từ đích về gốc
    path = []
    if path_found:
        curr = end
        while curr is not None:
            path.append(curr)
            curr = parent[curr]
        path.reverse()
        
    return final_cost, path

result_ucs = fixed_UCS(graph_2, start_2, goal_2)
print("Kết quả sử dụng thuật toán UCS: \n", result_ucs)
```
